# Introduction to an Array.

An array is a data structure that is designed for storing a collection of data in a sequential order.

```python
family = ['Father','Mother','Son','Daughter']
```

Arrays are numbered sequentially, starting with zero.

Arrays are one of the most popular data structure that many other data structure will use as part of their functioning.

Operations on Array:


1.   Traversal
2.   Access
3.   Insertion
4.   Deletion
5.   Searching
6.   Updating
7.   Sorting



Python don't have arrays they use list.

# Traversal of an Array.

Traversal of array:

```python
for i in family:
  print(i, end=" ")
```

# Accessing an Element from the Array.

Accessing an element:

```python
print(family[2])
```

# Adding element in an Array.

Adding an element : in between, at last and at beginning.

```python
family = ['Father','Mother','Son','Daughter']
```

Adding an element at the last of an Array.

```python
family.append("Grandfather")
```

```python
for i in family:
  print(i, end=" ")
```

Adding an element in the middle of an Array.

```python
family.insert(4,"Grandmother")
```

```python
for i in family:
  print(i, end=" ")
```

Adding an element at the beginning of the Array.

```python
family.insert(0,"Grandson")
```

```python
for i in family:
  print(i, end=" ")
```
