Linked List:

They provide an elegant solution for dealing with large amount of data that are constantly changing.

Linked list just like arrays helps us to store a collection of data.

Linked list work on relying on individual nodes that are connected to each other.

Each node is responsible for two things :

1.   What ever data is storing.
2.   A next pointer that points to the next node.



Linked lists were created to overcome various drawbacks associated with storing data in the array.
