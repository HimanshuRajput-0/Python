Linked List:

They provide an elegant solution for dealing with large amount of data that are constantly changing.

Linked list just like arrays helps us to store a collection of data.

Linked list work on relying on individual nodes that are connected to each other.

Each node is responsible for two things :

1.   What ever data is storing.
2.   A next pointer that points to the next node.



Linked lists were created to overcome various drawbacks associated with storing data in the array.

# Creation a singly linked list:

As linked lists consists of nodes so we will start with the creation of single node.

```python
class node:
  def __init__(self,data):
    self.data = data
    self.next = None
```

As now we have node so can easily make a singly linked linked-list.

first i will create multiple nodes then i will connect them with each other singly.

```python
n1 = node(10)
n2 = node(20)
n3 = node(30)
n4 = node(40)
```

Now its time to connect them singly using next pointer.

```python
n1.next = n2
n2.next = n3
n3.next = n4
```

Here we just linked all the nodes with eachother. Now we need to define the Header of our list.

```python
Header = n1
```

So we successfully created a single linked list.

# Traversing a singly linked list:

```python
current = Header
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```

# Insertion of a node at the beginning:

```python
def insert_at_beginning(Header, data):
    new_node = node(data)
    new_node.next = Header
    Header = new_node
    return Header
```

```python
Header = insert_at_beginning(Header, 5)
```

```python
current = Header
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```
