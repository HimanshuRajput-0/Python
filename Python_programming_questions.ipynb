# High Level Questions:

Write a program to solve a classic ancient Chinese puzzle: We count 35 heads and 94 legs among the chickens and rabbits in a farm. How many rabbits and how many chickens do we have?

Solution:

```python
def rabbit_chick(heads, legs):
  rabbits = (legs - 2*heads)/2
  chickens = heads - rabbits
  return rabbits, chickens
```

```python
heads = 35
legs = 94
rabbit_chick(heads, legs)
```

Write a program which prints all permutations of [1,2,3]

Solution:

```python
def permutation(array, start):
  if start == len(array):
     print(array)
     return

  for i in range(start, len(array)):
    array[start], array[i] = array[i], array[start]

    permutation(array, start+1)
    array[start], array[i] = array[i], array[start]
```

```python
array = [1, 2, 3]
permutation(array, 0)
```

Question 1: Find all the numbers which are divisible by 7 but not by 5 between 500 to 700.

The numbers obtained should be printed on a comma seperated sequence.

```python
list = []

for i in range(500,700):
  if i%7 == 0 and i%5 != 0:
    list.append(i)

print(list)
```

Question 2: Write a program to compute the factorial of a number.

```python
class factorialofnum:
  def factorial(self, num):
    fac = 1
    while num > 0:
      fac = fac*num
      num = num-1
    print(fac)
```

```python
facnum = factorialofnum()
facnum.factorial(5)
```

# Pattern Printing Questions:


Question 1: Half Pyramid.

```python
n= 5
for i in range(1, n+1):
  for j in range(1, i+1):
    print("*", end = " ")
  print() #this is empty to make the code know to start a new line after every row.
```

Question 2: Inverted half pyramid.

```python
n = 5
for i in range(1, n+1):
  for j in range(n, i-1, -1):
    print("*", end = " ")
  print()
```
